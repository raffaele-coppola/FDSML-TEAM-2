```python
from config import models, param_grids
from nested_cv import NestedCV
from scipy.stats import uniform
import numpy as np
from utils import get_cleaned_dataset
from steps import preprocessing_pipeline
```

```python
df = get_cleaned_dataset()
X = df.drop(columns=['target', 'diagnosis'])
y = df['target']
```

```python
params_to_add = {
    "grid": {
        "welch__percentage": np.linspace(0.1, 1.0, 10)
    },
    "random": {
        "welch__percentage": uniform(0.1, 0.9)
    }
}
```

```python
nested_cv = NestedCV(
    X=X,
    y=y,
    n_outer=5,
    n_inner=5,
    n_random_iter=60,
    scoring="recall"
)
```

```python
results = nested_cv.batch_run(
    pipeline=preprocessing_pipeline,
    models=models,
    model_params=param_grids,
    params_to_add=params_to_add,
    save_results=True,
    output_format="pickle"
)
```
